```python
intensities = [-1, 0, 127, 128, 255, 256]

for val in intensities:
    if val < 0 or val > 255:
        print(f"Intensity {val}: Invalid")
    elif val >= 128:
        print(f"Intensity {val}: Bright pixel")
    else:
        print(f"Intensity {val}: Dark pixel")
```

```python
pixel_values = [20, 130, 255]
for p in pixel_values:
    if p >= 128:
        print(f"Value {p}: Bright pixel")
    else:
        print(f"Value {p}: Dark pixel")

i = 0
while i < 3:
    print(i)
    i += 1
```

```python
def is_bright(pixel, threshold=128):
    if pixel >= threshold:
        return True
    else:
        return False

print(is_bright(127))       # Expected: False
print(is_bright(128))       # Expected: True
print(is_bright(200, 210))  # Expected: False
```
